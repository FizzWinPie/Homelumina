**SECTION 1: Pre-process for US Household mean income data from the US Census Database**

This is setting up the path for the data file 'data_path' is the variable. Then, run the cell to save your variables. PLEASE UPDATE THE DATA PATH TO REFLECT WHERE YOU STORE THE PRE DATA

```python
data_path = '/content/drive/MyDrive/CIS550_Project/salary.csv' # Path to life expectancy dataset in your drive
```

Setting up the pandas dataframe. We will take a quick overview of the first 10 rows

```python
df = pd.read_csv(data_path)
df.head(10)
```

We will drop the first row as it is additional information on the column name. We will also kept the relevant columns

```python
df = df.drop(0)
kept_col = ["GEO_ID", "NAME", "S1902_C03_001E"]
df = df[kept_col]
```

I save the dataframe to a separate DF to be manipulated.
We will print the row where income is not numeric. The output of the new dataframe shows that we have 1853 row of no income value

```python
modified_data = df
modified_data[~modified_data["S1902_C03_001E"].str.isnumeric()]
```

We will first change the name of the column to make it more readable.

```python
modified_data.rename(columns={'GEO_ID': 'GeoID', 'NAME': 'ZipCode', 'S1902_C03_001E': 'MeanIncome'}, inplace=True)
```

We will parse the 5 unit of digit for the zip code by using regex

```python
modified_data['ZipCode'] = modified_data['ZipCode'].str.extract(r'(\d{5})')
```

We will take a quick look of the dataframe here

```python
modified_data
```

We still have some row where MeanIncome is non-numeric. We will update these values to be Null

```python
modified_data['MeanIncome'] = pd.to_numeric(modified_data['MeanIncome'], errors='coerce')
modified_data
```

We will export the completed dataFrame to the path listed here

```python
modified_data.to_csv('/content/drive/MyDrive/CIS550_Project/clean_mean_income.csv', index=False)
```

**SECTION 2: Pre-process for Housing Data from Realtor.com**

This is the data path for the Realtor dataset. Please update your file path accordingly

```python
data_path = '/content/drive/MyDrive/RDC_Inventory_Core_Metrics_Zip_History.csv'
```

Setting up the pandas dataframe. We will take a quick overview of the first 10 rows

```python
house = pd.read_csv(data_path)
house.head(10)
```

We will kept only the relevant column

```python
kept_col = ["month_date_yyyymm", "postal_code", "median_listing_price", "active_listing_count"]
house = house[kept_col]
```

We will first change the name of the column to make it more readable.

```python
house.rename(columns={'month_date_yyyymm': 'MonthDate', 'postal_code': 'ZipCode', 'median_listing_price': 'MedianListingPrice', 'active_listing_count': 'ActiveListingCount'}, inplace=True)
```

We will check the type of each column in the Dataframe

```python
house.dtypes
```

We will convert the MonthDate to be str type as well

```python
house['MonthDate'] = house['MonthDate'].astype(str)
```

We will convert the zipcode column to be str type

```python
house['ZipCode'] = house['ZipCode'].astype(str)
```

We will use zfill to pad the zipCode to be 5 digit.

```python
house['ZipCode'] = house['ZipCode'].str.zfill(5)
```

```python
house.head(20)
```

Key attribute contain no null. We will export the current dataframe

```python
house.isnull().any()
```

We will export the dataframe to csv

```python
house.to_csv('/content/drive/MyDrive/CIS550_Project/clean_realtor.csv', index=False)
```
